```python
import scanpy as sc

# ===== 修改 celltype 的函数 =====
def assign_celltype_from_key(adatas_dict):
    
    for key, ad in adatas_dict.items():
        
        # key 形式例如：
        # merged_pre_res_HSCLSC
        # unmerged_post_nres_MPPCLP
        
        celltype = key.split("_")[-1]  # 取最后一个字段
        
        # 写入 obs
        ad.obs["celltype"] = celltype
        
        # 可选：设为 category
        ad.obs["celltype"] = ad.obs["celltype"].astype("category")
        
    return adatas_dict


# ===== 对 merged 数据修改 =====
merged_adatas = assign_celltype_from_key(merged_adatas)

# ===== 对 unmerged 数据修改 =====
unmerged_adatas = assign_celltype_from_key(unmerged_adatas)


# ===== 验证 =====
print("Merged check:")
for k, ad in merged_adatas.items():
    print(k, ad.obs["celltype"].unique())

print("\nUnmerged check:")
for k, ad in unmerged_adatas.items():
    print(k, ad.obs["celltype"].unique())
```

```python
import pandas as pd

records = []

# 处理 merged
for key, ad in merged_adatas.items():
    df = ad.obs.copy()
    df["cohort"] = "merged"
    df["group_key"] = key
    df["cell_id"] = df.index
    records.append(df)

# 处理 unmerged
for key, ad in unmerged_adatas.items():
    df = ad.obs.copy()
    df["cohort"] = "unmerged"
    df["group_key"] = key
    df["cell_id"] = df.index
    records.append(df)

# 合并
cell_table = pd.concat(records, axis=0)

# 只保留你关心的列（比如 celltype）
cell_table = cell_table[["cell_id", "cohort", "celltype", "group_key"]]

cell_table.to_csv("./comparison.csv")
```

```python
split_cols = cell_table["group_key"].str.split("_", expand=True)

cell_table["version"] = split_cols[0]                    # merged / unmerged
cell_table["cohort"]  = split_cols[1] + "_" + split_cols[2]  # pre_res / post_nres
cell_table["source_type"] = split_cols[3]                # MPPCLP 等

cell_table.head()
```

```python
wide = (
    cell_table
    .pivot_table(index=["cell_id", "cohort"],
                 columns="version",
                 values="celltype",
                 aggfunc="first")
    .reset_index()
)
wide["merged"] = wide["merged"].fillna("Absent")
wide["unmerged"] = wide["unmerged"].fillna("Absent")

for c in wide["cohort"].unique():
    
# for c in  ['post_nres', 'pre_nres']:  # Ensure we're working with the correct cohort names
    
    subset = wide[wide["cohort"] == c]
    
    crosstab = pd.crosstab(subset["unmerged"], subset["merged"])
    
    print(f"\n===== {c} =====")
    print(crosstab)
    # break
```

```python
import plotly.graph_objects as go
import pandas as pd

cohort_name = "pre_res"  # 改成你要画的 cohort

subset = wide[wide["cohort"] == cohort_name]

# 统计流向
flow = (
    subset
    .groupby(["unmerged", "merged"])
    .size()
    .reset_index(name="count")
)

# 左侧节点（unmerged）
left_labels = sorted(flow["unmerged"].unique())

# 右侧节点（merged）
right_labels = sorted(flow["merged"].unique())

# 合并节点列表（左在前，右在后）
labels = left_labels + right_labels

# 建立索引
label_to_index = {label: i for i, label in enumerate(labels)}

# 构建 source / target
source = flow["unmerged"].map(label_to_index)
target = flow["merged"].map(label_to_index)
value  = flow["count"]

fig = go.Figure(data=[go.Sankey(
    node=dict(
        pad=20,
        thickness=20,
        label=labels
    ),
    link=dict(
        source=source,
        target=target,
        value=value
    )
)])

fig.update_layout(title_text=f"Sankey - {cohort_name}", font_size=12)
fig.show()
```
